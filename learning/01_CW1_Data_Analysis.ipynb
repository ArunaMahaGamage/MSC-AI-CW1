## 2.2 Hidden Missing Values Assessment

### Objective

To identify text fields containing blank strings,
whitespace-only values and explicit placeholders
that are not necessarily detected by standard
null-value checks.

### Method

Text columns will be inspected without altering
the original dataset.

Whitespace-only values and explicit unknown-value
labels will be assessed separately.

### Importance

A column may have zero null values but still contain
unknown, unclassified or otherwise uninformative data.


```python
# Identify whitespace-only text values
text_columns = df_raw.select_dtypes(
    include=["object", "string", "str"]
)

whitespace_counts = text_columns.apply(
    lambda col: (
        col.astype("string")
        .str.strip()
        .eq("")
        .sum()
    )
)

whitespace_counts = whitespace_counts.sort_values(
    ascending=False
)

display(whitespace_counts)
```

```python
# Find explicit unknown-value labels
unknown_labels = {
    "UNKNOWN",
    "OTHER/UNKNOWN",
    "NOT REPORTED",
    "NOT PROVIDED",
    "MISSING",
    "UNSPECIFIED"
}

unknown_summary = {}

for column in text_columns.columns:

    normalized = (
        text_columns[column]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    matches = normalized.isin(unknown_labels)

    count = matches.sum()

    if count > 0:
        unknown_summary[column] = int(count)

unknown_summary = pd.Series(
    unknown_summary,
    name="Explicit Unknown Labels"
).sort_values(ascending=False)

display(unknown_summary)
```

```python
#  Inspect the Place category distribution
place_counts = df_raw["Place"].value_counts(
    dropna=False
)

display(place_counts.head(20))

unknown_place_count = (
    df_raw["Place"]
    .astype("string")
    .str.strip()
    .str.upper()
    .eq("OTHER/UNKNOWN")
    .sum()
)

unknown_place_pct = (
    unknown_place_count / len(df_raw)
) * 100

print("Other/Unknown records:", unknown_place_count)
print(
    "Other/Unknown percentage:",
    round(unknown_place_pct, 2),
    "%"
)
```

## 2.4 Date and Time Quality Assessment

### Objective

To evaluate the completeness, validity and chronological
consistency of the date/time fields in the crime dataset.

### Fields

- Dispatch Date / Time
- Start_Date_Time
- End_Date_Time

### Method

1. Preserve the original dataset.
2. Convert date strings into datetime values.
3. Quantify parsing failures and missing timestamps.
4. Identify records where the end time precedes the start time.
5. Identify records outside the stated study period.

### Importance

Invalid or inconsistent timestamps may distort
temporal trends, seasonal patterns and duration
calculations.

No records will be removed during the initial audit.


```python

# B4.2 - Create a working copy

df_dates = df_raw.copy()

date_columns = [
    "Dispatch Date / Time",
    "Start_Date_Time",
    "End_Date_Time"
]

date_format = "%m/%d/%Y %I:%M:%S %p"

# Convert dates and measure parsing failures
for column in date_columns:

    original = df_raw[column]

    converted = pd.to_datetime(
        original,
        format=date_format,
        errors="coerce"
    )

    parsing_failures = (
        original.notna() & converted.isna()
    ).sum()

    df_dates[column] = converted

    print(f"{column}")
    print(f"  Parsing failures: {parsing_failures}")
    print(f"  Missing after conversion: {converted.isna().sum()}")
    print()
```

```python

# B4.3 - Identify end times earlier than start times

invalid_end_before_start = (
    df_dates["End_Date_Time"].notna()
    & df_dates["Start_Date_Time"].notna()
    & (
        df_dates["End_Date_Time"]
        < df_dates["Start_Date_Time"]
    )
)

invalid_end_count = invalid_end_before_start.sum()

print(
    "End time before start time:",
    invalid_end_count
)

print(
    "Percentage of all records:",
    round(
        invalid_end_count / len(df_dates) * 100,
        2
    ),
    "%"
)

display(
    df_dates.loc[
        invalid_end_before_start,
        [
            "Incident ID",
            "Start_Date_Time",
            "End_Date_Time",
            "Crime Name2"
        ]
    ].head(10)
)
```

```python

# B4.4 - Check dates outside the stated coursework period

study_start = pd.Timestamp("2016-07-01 00:00:00")
study_end = pd.Timestamp("2026-09-10 23:59:59")

start_dates = df_dates["Start_Date_Time"]

outside_period = (
    (start_dates < study_start)
    | (start_dates > study_end)
)

print(
    "Records outside stated study period:",
    outside_period.sum()
)

print(
    "Earliest recorded start time:",
    start_dates.min()
)

print(
    "Latest recorded start time:",
    start_dates.max()
)

display(
    df_dates.loc[
        outside_period,
        ["Incident ID", "Start_Date_Time", "Crime Name2"]
    ].head(10)
)
```


### 2.4.1 Investigation of Chronological Inconsistencies

The initial audit identified 8,418 records where
End_Date_Time precedes Start_Date_Time.

Several inspected records contain an end timestamp
of exactly midnight (00:00:00).

This may indicate a reporting convention, a default
timestamp or a genuine recorded time.

Further analysis is required to determine how
frequently this pattern occurs.

No timestamps will be corrected or removed during
this investigation.


```python

# B4.5: Investigate midnight end timestamps

midnight_end = (
    df_dates["End_Date_Time"].notna()
    & df_dates["End_Date_Time"].dt.hour.eq(0)
    & df_dates["End_Date_Time"].dt.minute.eq(0)
    & df_dates["End_Date_Time"].dt.second.eq(0)
)

invalid_and_midnight = (
    invalid_end_before_start & midnight_end
)

invalid_not_midnight = (
    invalid_end_before_start & ~midnight_end
)

print("Total records with midnight End Time:",
      midnight_end.sum())

print("End-before-start with midnight End Time:",
      invalid_and_midnight.sum())

print("End-before-start with non-midnight End Time:",
      invalid_not_midnight.sum())

print(
    "Percentage of inconsistent records with midnight End Time:",
    round(
        invalid_and_midnight.sum()
        / invalid_end_before_start.sum() * 100,
        2
    )
)
```

```python

# B5.3 - Check missing and zero coordinates

latitude = pd.to_numeric(
    df_raw["Latitude"],
    errors="coerce"
)

longitude = pd.to_numeric(
    df_raw["Longitude"],
    errors="coerce"
)

missing_coordinates = (
    latitude.isna() | longitude.isna()
)

zero_coordinates = (
    latitude.eq(0) | longitude.eq(0)
)

print("Total records:", len(df_raw))

print(
    "Records with missing coordinates:",
    missing_coordinates.sum()
)

print(
    "Records with zero latitude or longitude:",
    zero_coordinates.sum()
)

print(
    "Zero coordinate percentage:",
    round(
        zero_coordinates.mean() * 100,
        2
    ),
    "%"
)
```

```python

# B5.4 - Validate coordinate ranges

invalid_latitude = (
    (latitude < -90) | (latitude > 90)
)

invalid_longitude = (
    (longitude < -180) | (longitude > 180)
)

invalid_coordinate_range = (
    invalid_latitude | invalid_longitude
)

print(
    "Invalid latitude values:",
    invalid_latitude.sum()
)

print(
    "Invalid longitude values:",
    invalid_longitude.sum()
)

print(
    "Records outside valid global coordinate ranges:",
    invalid_coordinate_range.sum()
)
```

**Important Consideration**

A statistical outlier does not necessarily indicate

The Victims field is analysed at the offence-row

```python

# B6.3 - Analyse the Victims column

victims = pd.to_numeric(
    df_raw["Victims"],
    errors="coerce"
)

print("Total records:", len(victims))
print("Missing victim values:", victims.isna().sum())
print("Negative victim values:", (victims < 0).sum())
print("Zero victim values:", (victims == 0).sum())

print("\nDescriptive statistics:")
display(victims.describe())

print("\nMost common victim counts:")
display(victims.value_counts().head(15))
```

```python

# B6.4 - Detect potential outliers using IQR

q1 = victims.quantile(0.25)
q3 = victims.quantile(0.75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

potential_outliers = (
    (victims < lower_bound)
    | (victims > upper_bound)
)

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

print(
    "Potential outlier records:",
    potential_outliers.sum()
)

print(
    "Potential outlier percentage:",
    round(potential_outliers.mean() * 100, 2),
    "%"
)
```
