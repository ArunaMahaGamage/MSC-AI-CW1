## 2.2 Hidden Missing Values Assessment

### Objective

To identify text fields containing blank strings,
whitespace-only values and explicit placeholders
that are not necessarily detected by standard
null-value checks.

### Method

Text columns will be inspected without altering
the original dataset.

Whitespace-only values and explicit unknown-value
labels will be assessed separately.

### Importance

A column may have zero null values but still contain
unknown, unclassified or otherwise uninformative data.


```python
# Identify whitespace-only text values
text_columns = df_raw.select_dtypes(
    include=["object", "string", "str"]
)

whitespace_counts = text_columns.apply(
    lambda col: (
        col.astype("string")
        .str.strip()
        .eq("")
        .sum()
    )
)

whitespace_counts = whitespace_counts.sort_values(
    ascending=False
)

display(whitespace_counts)
```

```python
# Find explicit unknown-value labels
unknown_labels = {
    "UNKNOWN",
    "OTHER/UNKNOWN",
    "NOT REPORTED",
    "NOT PROVIDED",
    "MISSING",
    "UNSPECIFIED"
}

unknown_summary = {}

for column in text_columns.columns:

    normalized = (
        text_columns[column]
        .astype("string")
        .str.strip()
        .str.upper()
    )

    matches = normalized.isin(unknown_labels)

    count = matches.sum()

    if count > 0:
        unknown_summary[column] = int(count)

unknown_summary = pd.Series(
    unknown_summary,
    name="Explicit Unknown Labels"
).sort_values(ascending=False)

display(unknown_summary)
```

```python
#  Inspect the Place category distribution
place_counts = df_raw["Place"].value_counts(
    dropna=False
)

display(place_counts.head(20))

unknown_place_count = (
    df_raw["Place"]
    .astype("string")
    .str.strip()
    .str.upper()
    .eq("OTHER/UNKNOWN")
    .sum()
)

unknown_place_pct = (
    unknown_place_count / len(df_raw)
) * 100

print("Other/Unknown records:", unknown_place_count)
print(
    "Other/Unknown percentage:",
    round(unknown_place_pct, 2),
    "%"
)
```

## 2.4 Date and Time Quality Assessment

### Objective

To evaluate the completeness, validity and chronological
consistency of the date/time fields in the crime dataset.

### Fields

- Dispatch Date / Time
- Start_Date_Time
- End_Date_Time

### Method

1. Preserve the original dataset.
2. Convert date strings into datetime values.
3. Quantify parsing failures and missing timestamps.
4. Identify records where the end time precedes the start time.
5. Identify records outside the stated study period.

### Importance

Invalid or inconsistent timestamps may distort
temporal trends, seasonal patterns and duration
calculations.

No records will be removed during the initial audit.


```python

# B4.2 - Create a working copy

df_dates = df_raw.copy()

date_columns = [
    "Dispatch Date / Time",
    "Start_Date_Time",
    "End_Date_Time"
]

date_format = "%m/%d/%Y %I:%M:%S %p"

# Convert dates and measure parsing failures
for column in date_columns:

    original = df_raw[column]

    converted = pd.to_datetime(
        original,
        format=date_format,
        errors="coerce"
    )

    parsing_failures = (
        original.notna() & converted.isna()
    ).sum()

    df_dates[column] = converted

    print(f"{column}")
    print(f"  Parsing failures: {parsing_failures}")
    print(f"  Missing after conversion: {converted.isna().sum()}")
    print()
```

```python

# B4.3 - Identify end times earlier than start times

invalid_end_before_start = (
    df_dates["End_Date_Time"].notna()
    & df_dates["Start_Date_Time"].notna()
    & (
        df_dates["End_Date_Time"]
        < df_dates["Start_Date_Time"]
    )
)

invalid_end_count = invalid_end_before_start.sum()

print(
    "End time before start time:",
    invalid_end_count
)

print(
    "Percentage of all records:",
    round(
        invalid_end_count / len(df_dates) * 100,
        2
    ),
    "%"
)

display(
    df_dates.loc[
        invalid_end_before_start,
        [
            "Incident ID",
            "Start_Date_Time",
            "End_Date_Time",
            "Crime Name2"
        ]
    ].head(10)
)
```

```python

# B4.4 - Check dates outside the stated coursework period

study_start = pd.Timestamp("2016-07-01 00:00:00")
study_end = pd.Timestamp("2026-09-10 23:59:59")

start_dates = df_dates["Start_Date_Time"]

outside_period = (
    (start_dates < study_start)
    | (start_dates > study_end)
)

print(
    "Records outside stated study period:",
    outside_period.sum()
)

print(
    "Earliest recorded start time:",
    start_dates.min()
)

print(
    "Latest recorded start time:",
    start_dates.max()
)

display(
    df_dates.loc[
        outside_period,
        ["Incident ID", "Start_Date_Time", "Crime Name2"]
    ].head(10)
)
```


### 2.4.1 Investigation of Chronological Inconsistencies

of exactly midnight (00:00:00).

```python

# B4.5: Investigate midnight end timestamps

midnight_end = (
    df_dates["End_Date_Time"].notna()
    & df_dates["End_Date_Time"].dt.hour.eq(0)
    & df_dates["End_Date_Time"].dt.minute.eq(0)
    & df_dates["End_Date_Time"].dt.second.eq(0)
)

invalid_and_midnight = (
    invalid_end_before_start & midnight_end
)

invalid_not_midnight = (
    invalid_end_before_start & ~midnight_end
)

print("Total records with midnight End Time:",
      midnight_end.sum())

print("End-before-start with midnight End Time:",
      invalid_and_midnight.sum())

print("End-before-start with non-midnight End Time:",
      invalid_not_midnight.sum())

print(
    "Percentage of inconsistent records with midnight End Time:",
    round(
        invalid_and_midnight.sum()
        / invalid_end_before_start.sum() * 100,
        2
    )
)
```

```python

# B5.3 - Check missing and zero coordinates

latitude = pd.to_numeric(
    df_raw["Latitude"],
    errors="coerce"
)

longitude = pd.to_numeric(
    df_raw["Longitude"],
    errors="coerce"
)

missing_coordinates = (
    latitude.isna() | longitude.isna()
)

zero_coordinates = (
    latitude.eq(0) | longitude.eq(0)
)

print("Total records:", len(df_raw))

print(
    "Records with missing coordinates:",
    missing_coordinates.sum()
)

print(
    "Records with zero latitude or longitude:",
    zero_coordinates.sum()
)

print(
    "Zero coordinate percentage:",
    round(
        zero_coordinates.mean() * 100,
        2
    ),
    "%"
)
```

```python

# RQ1.5 - Calculate Lift for the top 10 crime pairs

total_incidents = len(incident_categories)

# Count incidents containing each crime category
category_incident_counts = (
    df_group[["Incident ID", "Crime Name 2"]]
    .drop_duplicates()
    ["Crime Name 2"]
    .value_counts()
)

results = []

for pair, together_count in pair_counts.head(10).items():

    category_a, category_b = pair

    count_a = category_incident_counts[category_a]
    count_b = category_incident_counts[category_b]

    probability_a = count_a / total_incidents
    probability_b = count_b / total_incidents

    probability_together = (
        together_count / total_incidents
    )

    lift = probability_together / (
        probability_a * probability_b
    )

    results.append({
        "Category A": category_a,
        "Category B": category_b,
        "Together Count": together_count,
        "Category A Incidents": count_a,
        "Category B Incidents": count_b,
        "Lift": round(lift, 2)
    })

lift_results = pd.DataFrame(results)

display(
    lift_results.sort_values(
        "Lift",
        ascending=False
    )
)
```

The most frequent pair was Property Damage and
Theft From Motor Vehicle, with 2,664 incidents.

This shows that the most frequent pair does not

### 4.1.7 Contingency Table

I will create a contingency table for Drug Equipment
Violations and Drug/Narcotic Violations.

The table will show how many incidents contain both
categories, only one category, or neither category.

I will use this table to understand the relationship
between the two crime categories before carrying
out a statistical test.


```python

# RQ1.7 - Create a 2x2 contingency table

category_a = "Drug Equipment Violations"
category_b = "Drug/Narcotic Violations"

total = len(incident_categories)

count_a = category_incident_counts[category_a]
count_b = category_incident_counts[category_b]

both = pair_counts[
    tuple(sorted([category_a, category_b]))
]

only_a = count_a - both
only_b = count_b - both
neither = total - (both + only_a + only_b)

contingency_table = pd.DataFrame(
    {
        "Drug/Narcotic Yes": [both, only_b],
        "Drug/Narcotic No": [only_a, neither]
    },
    index=[
        "Drug Equipment Yes",
        "Drug Equipment No"
    ]
)

display(contingency_table)

print(
    "Total incidents in table:",
    contingency_table.to_numpy().sum()
)

print("Total incidents in dataset:", total)
```


### 4.1.8 Expected Frequencies

I will calculate the expected incident counts
using the contingency table.

The expected counts represent how the incidents
would be distributed if the two crime categories
were independent.

I will compare these values with the actual
counts before calculating the Chi-Square statistic.


```python

# RQ1.8 - Calculate expected frequencies

# Calculate row totals
row_totals = contingency_table.sum(axis=1)

# Calculate column totals
column_totals = contingency_table.sum(axis=0)

# Calculate total incidents
grand_total = contingency_table.to_numpy().sum()

# Create an empty table
expected_table = pd.DataFrame(
    index=contingency_table.index,
    columns=contingency_table.columns,
    dtype=float
)

# Calculate expected frequency for each cell
for row in contingency_table.index:
    for column in contingency_table.columns:

        expected = (
            row_totals[row] * column_totals[column]
        ) / grand_total

        expected_table.loc[row, column] = expected

print("Observed frequencies:")
display(contingency_table)

print("\nExpected frequencies:")
display(expected_table.round(2))

print(
    "\nExpected total:",
    round(expected_table.to_numpy().sum(), 2)
)
```


### 4.1.9 Chi-Square Calculation

I calculated the expected counts for the four
combinations of Drug Equipment Violations and
Drug/Narcotic Violations.

Next, I will calculate the Chi-Square statistic
using the observed and expected frequencies.

This will help me measure how much the actual
counts differ from the counts expected if the
two categories were independent.


```python

# RQ1.9 - Calculate Chi-Square statistic

# Difference between observed and expected counts
difference = contingency_table - expected_table

# Square the differences
squared_difference = difference ** 2

# Calculate the contribution from each cell
chi_square_contributions = (
    squared_difference / expected_table
)

print("Chi-Square contribution from each cell:")
display(chi_square_contributions.round(2))

# Add all four contributions
chi_square_statistic = (
    chi_square_contributions.to_numpy().sum()
)

# Degrees of freedom
rows, columns = contingency_table.shape

degrees_of_freedom = (rows - 1) * (columns - 1)

print(
    "\nChi-Square statistic:",
    round(chi_square_statistic, 2)
)

print(
    "Degrees of freedom:",
    degrees_of_freedom
)
```


### 4.1.10 Hypothesis Decision

I will compare the calculated Chi-Square statistic
with the critical value for a 5% significance level.

The test has one degree of freedom because the
contingency table contains two rows and two columns.

This comparison will help me decide whether the
results provide enough evidence to reject the
null hypothesis.


```python

# RQ1.10 - Chi-Square hypothesis decision

significance_level = 0.05

# Reference value for df = 1 and alpha = 0.05
critical_value = 3.841

print("Significance level:", significance_level)
print("Degrees of freedom:", degrees_of_freedom)
print("Critical value:", critical_value)

print(
    "Calculated Chi-Square:",
    round(chi_square_statistic, 2)
)

if chi_square_statistic > critical_value:
    print("\nDecision: Reject H0")
    print("There is evidence of an association.")
else:
    print("\nDecision: Fail to reject H0")
    print("There is insufficient evidence of an association.")
```


### 4.1.10 Hypothesis Test Result

The calculated Chi-Square statistic was 14,253.46
with one degree of freedom.

At the 5% significance level, the critical value
was 3.841.

Since the calculated value was greater than the
critical value, I rejected the null hypothesis.

The results provide evidence of an association
between Drug Equipment Violations and
Drug/Narcotic Violations in the recorded incidents.

This does not mean that one offence causes the other.

### 4.1.15 Comparing Crime Associations by Police District

I will calculate the association between Drug
Equipment Violations and Drug/Narcotic Violations

For each district, I will calculate the number
of incidents containing each category and the

I will then compare the observed and expected

```python

# RQ1.15 - Calculate Lift for each police district

# Create a column indicating whether both categories are present
district_crime_data["Both Categories"] = (
    district_crime_data["Drug Equipment"]
    & district_crime_data["Drug/Narcotic"]
)

# Group by police district
district_summary = (
    district_crime_data.groupby("Police District Name")
    .agg(
        Total_Incidents=("Incident ID", "count"),
        Drug_Equipment=("Drug Equipment", "sum"),
        Drug_Narcotic=("Drug/Narcotic", "sum"),
        Both_Categories=("Both Categories", "sum")
    )
    .reset_index()
)

# Calculate expected co-occurrence
district_summary["Expected"] = (
    district_summary["Drug_Equipment"]
    * district_summary["Drug_Narcotic"]
    / district_summary["Total_Incidents"]
)

# Calculate Lift
district_summary["Lift"] = (
    district_summary["Both_Categories"]
    / district_summary["Expected"]
)

# Sort districts by Lift
district_summary = district_summary.sort_values(
    "Lift",
    ascending=False
)

print("Police district association results:")

display(district_summary.round(2))
```

```python

# RQ1.21 - Approximate Odds Ratio homogeneity test
# Using Pandas and NumPy only

import numpy as np

homogeneity_data = district_contingency.copy()

a = homogeneity_data["Both"]
b = homogeneity_data["Equipment Only"]
c = homogeneity_data["Narcotic Only"]
d = homogeneity_data["Neither"]

# Check that all four counts are positive
if (homogeneity_data[
    ["Both", "Equipment Only", "Narcotic Only", "Neither"]
] <= 0).any().any():
    raise ValueError("Zero or negative cell count found")

# Step 1 - Calculate log Odds Ratios
homogeneity_data["Log OR"] = np.log(
    (a * d) / (b * c)
)

# Step 2 - Estimate variance of each log OR
homogeneity_data["Variance"] = (
    1/a + 1/b + 1/c + 1/d
)

# Step 3 - Assign inverse-variance weights
homogeneity_data["Weight"] = (
    1 / homogeneity_data["Variance"]
)

# Step 4 - Calculate weighted average log OR
weighted_average = (
    (homogeneity_data["Weight"] *
     homogeneity_data["Log OR"]).sum()
    / homogeneity_data["Weight"].sum()
)

# Step 5 - Calculate heterogeneity statistic
homogeneity_data["Contribution"] = (
    homogeneity_data["Weight"]
    * (homogeneity_data["Log OR"] - weighted_average) ** 2
)

test_statistic = homogeneity_data["Contribution"].sum()

df = len(homogeneity_data) - 1

print("Districts:", len(homogeneity_data))
print("Test statistic:", round(test_statistic, 2))
print("Degrees of freedom:", df)

print("\nContributions by district:")
display(
    homogeneity_data[
        ["Police District Name", "Contribution"]
    ].round(2)
)

# Reference critical value for df = 6, alpha = 0.05
critical_value = 12.592

print("\nCritical value (5%):", critical_value)

if test_statistic > critical_value:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```


### 4.1.21 District Association Test Results

I compared the Odds Ratios across the seven
named police districts using an approximate
homogeneity test.

The test statistic was 98.44 with six degrees
of freedom. The critical value at the 5%
significance level was 12.592.

Since the test statistic was greater than
the critical value, I rejected the null
hypothesis.

This provides evidence that the association
between Drug Equipment Violations and
Drug/Narcotic Violations is not the same
across all seven districts.

Bethesda, Wheaton and Rockville made the
largest contributions to the test statistic.

The test does not show exactly which district
pairs differ significantly or explain why
the differences exist.




### 4.1.22 Exploring Other Crime Category Pairs

So far, I have calculated Lift for the ten

I will now examine other pairs that occur

This will help me identify associations
that may have been missed when looking

```python

# RQ1.22 - Calculate Lift for all pairs with count >= 100

minimum_count = 100

eligible_pairs = pair_counts[
    pair_counts >= minimum_count
]

all_pair_results = []

total_incidents = len(incident_categories)

for pair, together_count in eligible_pairs.items():

    category_a, category_b = pair

    count_a = category_incident_counts[category_a]
    count_b = category_incident_counts[category_b]

    expected = (
        count_a * count_b
    ) / total_incidents

    lift = together_count / expected

    all_pair_results.append({
        "Category A": category_a,
        "Category B": category_b,
        "Together Count": together_count,
        "Expected": expected,
        "Lift": lift
    })

all_pair_lift = pd.DataFrame(all_pair_results)

all_pair_lift = all_pair_lift.sort_values(
    "Lift",
    ascending=False
)

print("Pairs meeting minimum count:", len(all_pair_lift))

print("\nTop 15 pairs by Lift:")

display(all_pair_lift.head(15).round(2))
```

### 4.1.23 Checking the Highest-Lift Crime Pair

I will examine the two crime categories that
produced the highest Lift value.

I will check how many incidents contain each
category and how many contain both.

I will also calculate support and confidence
to understand the association more clearly.

This will help me decide how reliable and
meaningful the high Lift value is.


```python

# RQ1.23 - Investigate the highest-Lift pair

category_a = "Assisting or Promoting Prostitution"
category_b = "Prostitution"

total_incidents = len(incident_categories)

# Number of incidents containing each category
count_a = category_incident_counts[category_a]
count_b = category_incident_counts[category_b]

# Number of incidents containing both categories
both_count = pair_counts[
    tuple(sorted([category_a, category_b]))
]

# Calculate support
support_percentage = (
    both_count / total_incidents
) * 100

# Calculate confidence in both directions
confidence_a_to_b = (
    both_count / count_a
) * 100

confidence_b_to_a = (
    both_count / count_b
) * 100

# Calculate expected count and Lift
expected = (
    count_a * count_b
) / total_incidents

lift = both_count / expected

print("Total incidents:", total_incidents)
print("Category A incidents:", count_a)
print("Category B incidents:", count_b)
print("Both categories:", both_count)

print("\nSupport (%):", round(support_percentage, 4))
print("Confidence A -> B (%):", round(confidence_a_to_b, 2))
print("Confidence B -> A (%):", round(confidence_b_to_a, 2))

print("\nExpected together:", round(expected, 4))
print("Lift:", round(lift, 2))
```

Both categories appeared together in 112
incidents.

The confidence from Assisting or Promoting
Prostitution to Prostitution was 72.73%.

However, the pair represented only 0.024%
of all incidents.

The Lift was very high (1176.18), partly
because the expected count was very small.

This shows why I need to consider support
and confidence together with Lift when
interpreting crime-category associations.



### 4.1.24 Checking Different Minimum Pair Counts

I used a minimum of 100 incidents when
selecting crime pairs for the Lift analysis.

I will compare this with other minimum
counts to see how the results change.

I will check how many pairs are included
at each threshold and which pair has
the highest Lift.

This will help me understand how the
selection threshold affects the analysis.


```python

# RQ1.24 - Sensitivity analysis for minimum pair counts

thresholds = [50, 100, 200, 500]

sensitivity_results = []

for minimum_count in thresholds:

    selected_pairs = pair_counts[
        pair_counts >= minimum_count
    ]

    pair_results = []

    for pair, together_count in selected_pairs.items():

        category_a, category_b = pair

        count_a = category_incident_counts[category_a]
        count_b = category_incident_counts[category_b]

        expected = (
            count_a * count_b
        ) / len(incident_categories)

        lift = together_count / expected

        pair_results.append({
            "Pair": pair,
            "Count": together_count,
            "Lift": lift
        })

    results_df = pd.DataFrame(pair_results)

    if len(results_df) > 0:

        highest = results_df.sort_values(
            "Lift",
            ascending=False
        ).iloc[0]

        sensitivity_results.append({
            "Minimum Count": minimum_count,
            "Eligible Pairs": len(results_df),
            "Highest Pair": str(highest["Pair"]),
            "Highest Pair Count": highest["Count"],
            "Highest Lift": highest["Lift"]
        })

sensitivity_table = pd.DataFrame(sensitivity_results)

print("Sensitivity analysis results:")
display(sensitivity_table.round(2))
```


### 4.1.24 Sensitivity Analysis Results

I compared the crime-pair results using minimum
counts of 50, 100, 200 and 500 incidents.

The number of eligible pairs decreased from
122 at a minimum count of 50 to 23 at 500.

The highest-Lift pair changed when I increased
the threshold.

At minimum counts of 200 and 500, Drug
Equipment Violations and Drug/Narcotic
Violations had the highest Lift (9.66).

This shows that the selection threshold affects
the results, especially for less common pairs.

I should consider both the number of incidents
and Lift when interpreting associations.
